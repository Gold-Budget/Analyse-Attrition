```python
import pandas as pd

cat_cols = df.select_dtypes(include='object').columns.tolist()
print("Colonnes catégorielles :", cat_cols)

df_encoded = pd.get_dummies(df, columns=cat_cols, drop_first=True)

print(df_encoded.shape)
df_encoded.head()

y = df_encoded['Churn']
X = df_encoded.drop(columns=['Churn'])

print(X.isnull().sum().sum())  

import statsmodels.api as sm

X_const = sm.add_constant(X)  
X_const = X_const.astype(float)  

model = sm.Logit(y, X_const).fit()
print(model.summary())

import numpy as np

resultats = pd.DataFrame({
    'coefficient': model.params,
    'odds_ratio': np.exp(model.params),
    'p_value': model.pvalues
}).drop('const')

resultats = resultats.sort_values('p_value')
print(resultats)

import matplotlib.pyplot as plt

significatifs = resultats[resultats['p_value'] < 0.05].sort_values('coefficient')

plt.figure(figsize=(9, max(4, len(significatifs) * 0.4)))
colors = ['crimson' if c > 0 else 'steelblue' for c in significatifs['coefficient']]
plt.barh(significatifs.index, significatifs['coefficient'], color=colors)
plt.axvline(0, color='black', linewidth=0.8)
plt.title("Variables ayant un effet significatif sur le churn\n(rouge = augmente le risque, bleu = le diminue)")
plt.xlabel("Coefficient (log-odds)")
plt.tight_layout()
plt.show()
```
